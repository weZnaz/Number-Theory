## 3. Mathematical background

Euler's totient function counts the integers from 1 to n that are relatively prime to n.

$$
arphi(n) = |\{k \in \{1,\ldots,n\}: \gcd(k,n)=1\}|
$$


## 4. Important equations

$$
 a^{arphi(n)} \equiv 1 \pmod n
$$

$$
 a^{p-1} \equiv 1 \pmod p
